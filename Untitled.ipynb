```python
import pandas as pd
from sqlalchemy import create_engine

# Connect to SQLite Northwind database
engine = create_engine('sqlite:///northwind.db')

# Example 1: Subset all customers from 'USA'
query_customers_usa = """
SELECT CustomerID, CompanyName, Country
FROM Customers
WHERE Country = 'USA';
"""

customers_usa = pd.read_sql(query_customers_usa, engine)
print(customers_usa.head())

# Example 2: Subset orders placed in 1997 with Freight > 50
query_orders_1997 = """
SELECT OrderID, CustomerID, OrderDate, Freight
FROM Orders
WHERE strftime('%Y', OrderDate) = '1997' AND Freight > 50;
"""

orders_1997 = pd.read_sql(query_orders_1997, engine)
print(orders_1997.head())

# Example 3: Join + subset: Products sold in category 'Beverages'
query_beverages = """
SELECT p.ProductID, p.ProductName, c.CategoryName, p.UnitPrice
FROM Products p
JOIN Categories c ON p.CategoryID = c.CategoryID
WHERE c.CategoryName = 'Beverages';
"""

beverages = pd.read_sql(query_beverages, engine)
print(beverages.head())
```
